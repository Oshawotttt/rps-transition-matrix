# Build transition matrices

Loads pooled human counts and builds the 3x3, 3x9 and 9x9 matrices.

```python
from pathlib import Path

from rps_markov.markov import Counts, OutcomeTransitionMatrix, BasicTransitionMatrix, BotMatchTransitionMatrix
from rps_markov.payoff import PayoffMatrix

ROOT = Path.cwd() if (Path.cwd() / "data").exists() else Path.cwd().parent  # repo root or notebooks/

counts = Counts.load(ROOT / "data/processed/transition_matrices/N_rps_human_combined.csv")

outcome = OutcomeTransitionMatrix(counts)        # 3 x 9
basic = BasicTransitionMatrix(counts)            # 3 x 3
bot = BotMatchTransitionMatrix(outcome, PayoffMatrix())  # 9 x 9
```
